## Polymorphism

**Exercise:** Study and modify the code. Try changing values and extending the class.

```python
class Dog:
    def speak(self): print('Woof')
class Cat:
    def speak(self): print('Meow')
for a in [Dog(),Cat()]: a.speak()
```

## Abstraction

**Exercise:** Study and modify the code. Try changing values and extending the class.

```python
from abc import ABC,abstractmethod
class Shape(ABC):
    @abstractmethod
    def area(self):...
class Square(Shape):
    def __init__(self,s): self.s=s
    def area(self): return self.s*self.s
print(Square(5).area())
```

## Static/Class Methods

**Exercise:** Study and modify the code. Try changing values and extending the class.

```python
class Math:
    count=0
    @staticmethod
    def add(a,b): return a+b
    @classmethod
    def inc(cls): cls.count+=1
print(Math.add(2,3)); Math.inc(); print(Math.count)
```
