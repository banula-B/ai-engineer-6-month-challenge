# Let's Dive into Pandas Library in Python

Pandas is a library that used to Manipulate data in python.
first we have to import pandas library to our project


```python
import pandas as pd
```

# Series -----------------------------

```python
data = [12,33,54,21,47]
```

```python
# we can convert this list into series using pandas

series = pd.Series(data)  #this .Series is a Constructor, not a function

print(series)
```

In a series, data is listed with a index, we can change the index as what we want

```python
series = pd.Series(data , index=["a","b","c","d","e"])
print(series)
```

We can find the value of any index using .loc[] property

```python
series = pd.Series(data,index=["a","b","c","d","e"])

print(series.loc["a"])
```

We can change the value of any index using .loc property

```python
print("series with no changes")
print(series)

#we can change the value of a index
print("series with changed c value")
series.loc["c"] = 200

print(series)
```

After we names our index using alphabet, still we can find them by usin indexes

for that we can use .iloc[] property

```python
print(series.iloc[2])
```

Also we can access values in series 

```python
#Access values with a boundary
print(series[series<30])

print(series[series>20])

#Access direct values
print(series[series==12])
```

Add values to Series

```python
# we can add values to series by using index and value
series["f"]=40
print(series)
```

Series using a dictionary

```python
calories = {"day1": 420, "day2": 380, "day3": 390}

series = pd.Series(calories)

print(series)
```

```python
series=pd.Series(calories, index =["day1"])
print(series)
```

# DataFrames ------------------------

```python
#for dataframes, we have to give our data in a dictionary

data = {
  "calories": [420, 380, 390],
  "duration": [50, 40, 45]
}
```

```python
df = pd.DataFrame(data)
print(df)
```

Change indexing in DataFrame

```python
# we can change the index no same as in Series

df = pd.DataFrame(data, index=["day1","day2","day3"])
print(df)
```

```python
# get data using index
# we can get all the data available in given index

print(df.loc["day1"])
```

```python
# Using iloc property

print(df.iloc[2])
```

### Add new items to DataFrame

lets add new column

```python
df["Activity"] = ["Joggin","Running","Running"]
print(df)
```

lets add a new row

```python
new_row=pd.DataFrame([{"calories" : 400, "duration": 45, "Activity": "Joggin"}], index=["day4"])
df=pd.concat([df,new_row])

print(df)
```

Lets add new rows

```python
new_rows=pd.DataFrame([{"calories" : 380, "duration": 20, "Activity": "Swimming"},
                        {"calories" : 250, "duration": 30, "Activity": "Walking"},
                        {"calories" : 350, "duration": 45, "Activity": "Cycling"}
                        ], index=["day5","day6","day7"])
df=pd.concat([df,new_rows])

print(df)
```

### Selection by Rows

To make selection by Rows, we can set the index of a data to the name of data
for example, 

index of the first pokemon is Bulbasaur
index of the second pokemon is Ivysaur

```python
# Set the name as index
df = pd.read_csv("../data sets/pokemon.csv", index_col = "Name")

print(df)
```

Now we can select rows by name of pokemon

```python
select_r = df.loc["Pikachu"]
print(select_r)
```

we can get selected columns of selected row data

```python
select_r = df.loc["Pikachu", ["Type1","Legendary"]]
print(select_r)
```

get selected set of pokemons data

```python
selected = df.loc["Charizard": "Blastoise", ["Type1","Height"]]
print(selected)
```

You can select rows and columns by column ID too

```python
selected = df.iloc[0:11 , 1:3]
print(selected)
```
