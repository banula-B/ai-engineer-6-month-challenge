# Library Management Mini Project

```python
class Library:
    def __init__(self):
        self.books=[]

    def add_book(self, book):
        self.books.append({"title":book,"available":True})
        print(f"{book} added.")

    def display_books(self):
        print("\nLibrary Books:")
        for b in self.books:
            status="Available" if b["available"] else "Borrowed"
            print(f"- {b['title']} ({status})")

    def borrow_book(self, book):
        for b in self.books:
            if b["title"]==book:
                if b["available"]:
                    b["available"]=False
                    print(f"You borrowed '{book}'.")
                else:
                    print("Book is already borrowed.")
                return
        print("Book not found.")

    def return_book(self, book):
        for b in self.books:
            if b["title"]==book:
                if not b["available"]:
                    b["available"]=True
                    print(f"You returned '{book}'.")
                else:
                    print("Book was not borrowed.")
                return
        print("Book not found.")

lib=Library()
lib.add_book("Python Basics")
lib.add_book("Data Structures")
lib.add_book("Machine Learning")

lib.display_books()
lib.borrow_book("Python Basics")
lib.display_books()
lib.return_book("Python Basics")
lib.display_books()
```
