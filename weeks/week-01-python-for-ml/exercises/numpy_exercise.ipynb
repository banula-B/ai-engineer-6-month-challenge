## Exercises 3


Create the following NumPy array:

```python
scores = np.array([
    [78, 65, 90, 45, 82],
    [55, 88, 72, 91, 60],
    [95, 42, 76, 84, 69]
])
```

Complete the following tasks:

1. Calculate the average score of all students.
2. Find the highest score.
3. Find the lowest score.
4. Calculate the average score of each student.
5. Calculate the average score of each subject.
6. Find all scores below `50`.
7. Use `np.where()` to replace scores below `50` with `0`.
8. Find the position of the highest score using `np.argmax()`.

````python
scores = np.array([
    [78, 65, 90, 45, 82],
    [55, 88, 72, 91, 60],
    [95, 42, 76, 84, 69]
])
````

````python
# 1
average = np.mean(scores)

# 2
highest = np.max(scores)

# 3
lowest = np.min(scores)

# 4
student_averages = np.mean(scores, axis=1)

# 5
subject_averages = np.mean(scores, axis=0)

# 6
below_50 = scores < 50

# 7
updated_scores = np.where(scores < 50, 0, scores)

# 8
index = np.argmax(scores)



print(f"1) {average}")
print(f"2) {highest}")
print(f"3) {lowest}")
print(f"4) {student_averages}")
print(f"5) {subject_averages}")
print(f"6) {below_50}")
print(f"7) {updated_scores}")
print(f"8) {index}")
````
