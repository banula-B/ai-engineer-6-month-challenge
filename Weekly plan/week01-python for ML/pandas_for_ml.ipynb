# Series -----------------------------

```python
data = [12,33,54,21,47]
```

```python
# we can convert this list into series using pandas

series = pd.Series(data)  #this .Series is a Constructor, not a function

print(series)
```

In a series, data is listed with a index, we can change the index as what we want

```python
series = pd.Series(data , index=["a","b","c","d","e"])
print(series)
```

We can find the value of any index using .loc[] property

```python
series = pd.Series(data,index=["a","b","c","d","e"])

print(series.loc["a"])
```

We can change the value of any index using .loc property

```python
print("series with no changes")
print(series)

#we can change the value of a index
print("series with changed c value")
series.loc["c"] = 200

print(series)
```

After we names our index using alphabet, still we can find them by usin indexes

for that we can use .iloc[] property

```python
print(series.iloc[2])
```

Also we can access values in series 

```python
#Access values with a boundary
print(series[series<30])

print(series[series>20])

#Access direct values
print(series[series==12])
```

Add values to Series

```python
# we can add values to series by using index and value
series["f"]=40
print(series)
```

Series using a dictionary

```python
calories = {"day1": 420, "day2": 380, "day3": 390}

series = pd.Series(calories)

print(series)
```

```python
series=pd.Series(calories, index =["day1"])
print(series)
```

# DataFrames ------------------------

```python
#for dataframes, we have to give our data in a dictionary

data = {
  "calories": [420, 380, 390],
  "duration": [50, 40, 45]
}
```

```python
df = pd.DataFrame(data)
print(df)
```

Change indexing in DataFrame

```python
# we can change the index no same as in Series

df = pd.DataFrame(data, index=["day1","day2","day3"])
print(df)
```

```python
# get data using index
# we can get all the data available in given index

print(df.loc["day1"])
```

```python
# Using iloc property

print(df.iloc[2])
```

# Filtering

Filtering ensure keeping the rows that match a condition.

With filtering, we can select filtered data as we need.
for example, we can select only students from school, height above 170cm

```python
# Let's select pokemons tall that 2 meters

tall_pokemon = df[df["Height"] >= 2]
print(tall_pokemon)
```

```python
# Let's select pokemons heavy than 100kg 

heavy_pokemon = df[df["Weight"]>100]

print(heavy_pokemon)
```

You can select items with using two or more conditions.

To filter data in pandas, you must use bitwise operators (&, |, ~) instead of standard Python logical operators (and, or, not)

```python
# Select item with two conditions

water_pokemon=df[(df["Type1"]== "Water") |
                (df["Type2"] == "Water")]

print(water_pokemon)
```

# Aggregation

Aggregation, reduces a set of values into a single summary value.

used to summerize and analyze data.

often used with the groupby() function

```python
# you can summary full table data

mean = df.mean(numeric_only=True)
print(mean)
```

In above code, we have put condition to the mean function. 
for pokemon names, we can't find a mean, sum, max, min or any mathematical values. so we have to neglect non integer rows in data. for that we use *numeric_only=True* condition

```python
sum = df.sum(numeric_only=True)
print(sum)
```

You can analyze data using one column data

```python
# Let's find the mean value of Heights of pokemons
sum_of_heigt=df["Height"].mean()

print(sum_of_heigt)
```

```python
mean_of_weight= df["Weight"].mean()

print(mean_of_weight)
```

We can group data in our dataset using any column data

```python
group=df.groupby("Type1")
print(group["Height"].mean())
```

# Data Cleaning

Drop irrelevant columns

```python
df = df.drop(columns=["Legendary", "No"])
print(df)
```

Handle missing data

```python
# Drop data with empty values 
# dropna = drop not available
df = df.dropna(subset=["Type2"])
print(df)
```

Standarize text

```python
# Let's set all the names written by lowercase only
df["Name"] = df["Name"].str.lower()
print(df)
```

Fix Data type

```python
# Let's assign boolean values for legendary type
df["Legendary"]= df["Legendary"].astype(bool)

print(df)
```

Remove duplicated values

```python
# You can remove duplicated values in our dataset

df = df.drop_duplicates()
print(df)
```

# Pivot 

A pivot table in Pandas is a powerful tool used to reshape, summarize, and aggregate tabular data into a clean, two-dimensional layout

```python
df=pd.read_csv("../data sets/weather.csv")
print(df)
```

```python
df.pivot(index="date", columns="city")
```

```python
# You can select only value for use

df.pivot(index="date", columns="city", values="temperature")
```
